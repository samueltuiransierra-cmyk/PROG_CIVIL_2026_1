### **Funciones Auxiliares para Validación y Conversión de Datos**
Esta celda contiene funciones auxiliares que se utilizan para validar y convertir tipos de datos (números o cadenas a flotantes) en los parámetros de entrada de las funciones principales, asegurando la robustez del código.

```python
def _validate_and_convert_numeric(value, name: str):
  """Validates if a value is numeric and converts it to float if possible."""
  if isinstance(value, (int, float)):
    return float(value)
  if isinstance(value, str):
    try:
      return float(value)
    except ValueError:
      raise TypeError(f"El parámetro '{name}' debe ser un número o una cadena convertible a número, pero se recibió '{value}' (tipo: {type(value).__name__}).")
  raise TypeError(f"El parámetro '{name}' debe ser un número o una cadena convertible a número, pero se recibió '{value}' (tipo: {type(value).__name__}).")

def _validate_and_convert_list_of_numerics(lst: list, name: str):
  """Validates if a list contains numeric values and converts them to float."""
  converted_list = []
  for i, item in enumerate(lst):
    converted_list.append(_validate_and_convert_numeric(item, f"{name}[{i}]"))
  return converted_list

def _validate_and_convert_dict_values(d: dict, name: str):
  """Validates if dictionary values are numeric and converts them to float."""
  converted_dict = {}
  for key, value in d.items():
    converted_dict[key] = _validate_and_convert_numeric(value, f"{name}['{key}']")
  return converted_dict
```

### **Función `volumen_absoluto`**
Esta función calcula el volumen absoluto (en m³) de los materiales de concreto a partir de sus masas (kg) y gravedades específicas, utilizando la fórmula $V = m / (SG \cdot 1000)$. Incluye validación de tamaño de listas.

```python
def volumen_absoluto(masses_kg: list, specific_gravities: list) -> list:
  """
  Calcula el volumen absoluto para una lista de masas y gravedades específicas.

  Args:
    masses_kg: Una lista de masas en kilogramos (kg).
    specific_gravities: Una lista de gravedades específicas (adimensional).

  Returns:
    Una lista de volúmenes en metros cúbicos (m³).

  Raises:
    ValueError: Si las listas de masas y gravedades específicas tienen diferente tamaño.
  """
  masses_kg = _validate_and_convert_list_of_numerics(masses_kg, 'masses_kg')
  specific_gravities = _validate_and_convert_list_of_numerics(specific_gravities, 'specific_gravities')

  if len(masses_kg) != len(specific_gravities):
    raise ValueError("Las listas de masas y gravedades específicas deben tener el mismo tamaño.")

  volumes_m3 = []
  for i in range(len(masses_kg)):
    mass = masses_kg[i]
    sg = specific_gravities[i]
    # Formula: V = m / (SG * 1000)
    volume = mass / (sg * 1000)
    volumes_m3.append(volume)

  return volumes_m3
```

### **Función `dosificar_concreto`**
Esta función calcula las masas finales de los componentes del concreto, escalándolas proporcionalmente al volumen total solicitado. También valida la relación agua/cemento calculada y emite una advertencia si excede un límite predefinido (0.65).

```python
def dosificar_concreto(
    f_c_mpa: float,
    total_volume_m3: float,
    ac_ratio: float, # Este parámetro se mantiene por contexto de diseño, pero la validación se hará con el ratio efectivo.
    unit_masses_per_m3: dict
) -> tuple[dict, str | None]:
  """
  Calcula las masas finales de los componentes del concreto escalando
  proporcionalmente al volumen pedido y valida la relación agua/cemento.

  Args:
    f_c_mpa: Resistencia a la compresión del concreto en MPa (f'c).
    total_volume_m3: Volumen total de concreto pedido en metros cúbicos (m³).
    ac_ratio: Relación agua/cemento deseada (para contexto de diseño).
    unit_masses_per_m3: Diccionario con las masas unitarias (kg) de cemento, agua,
                        arena y grava para 1 m³ de concreto.
                        Ejemplo: {'cement': 350, 'water': 175, 'sand': 800, 'gravel': 1000}

  Returns:
    Una tupla: (diccionario con las masas finales (kg) de cada componente,
                mensaje de advertencia (str) si la relación a/c excede 0.65, o None).

  Notes:
    Internamente, esta función utiliza 'volumen_absoluto' para calcular
    el volumen de cada componente, utilizando gravedades específicas típicas.
  """
  f_c_mpa = _validate_and_convert_numeric(f_c_mpa, 'f_c_mpa')
  total_volume_m3 = _validate_and_convert_numeric(total_volume_m3, 'total_volume_m3')
  ac_ratio = _validate_and_convert_numeric(ac_ratio, 'ac_ratio')
  unit_masses_per_m3 = _validate_and_convert_dict_values(unit_masses_per_m3, 'unit_masses_per_m3')

  final_masses = {}
  for component, unit_mass in unit_masses_per_m3.items():
    final_masses[component] = unit_mass * total_volume_m3

  warning_message = None

  # Validar la relación agua/cemento calculada
  if 'water' in final_masses and 'cement' in final_masses and final_masses['cement'] > 0:
    ac_ratio_calculated = final_masses['water'] / final_masses['cement']
    if ac_ratio_calculated > 0.65:
      warning_message = f"Advertencia: La relación agua/cemento calculada ({ac_ratio_calculated:.2f}) excede el límite de 0.65. Se recomienda ajustar la dosificación."

  # Para demostrar el uso interno de volumen_absoluto
  # Asumimos gravedades específicas típicas para los componentes del concreto.
  specific_gravities = {
      'cement': 3.15,
      'water': 1.0,
      'sand': 2.65,
      'gravel': 2.65
  }

  masses_for_vol_calc = []
  sgs_for_vol_calc = []
  component_names = []

  # Asegurarse de que el orden de componentes sea consistente
  ordered_components = ['cement', 'water', 'sand', 'gravel']

  for component in ordered_components:
    if component in final_masses and component in specific_gravities:
      masses_for_vol_calc.append(final_masses[component])
      sgs_for_vol_calc.append(specific_gravities[component])
      component_names.append(component)
    else:
      print(f"Advertencia: No se encontró masa final o gravedad específica para '{component}'. No se incluirá en el cálculo de volumen.")

  # Llama a volumen_absoluto internamente
  if masses_for_vol_calc and sgs_for_vol_calc:
    try:
      component_volumes_m3 = volumen_absoluto(masses_for_vol_calc, sgs_for_vol_calc)
      print(f"\nVolúmenes individuales de los componentes calculados internamente (m³):")
      for i, vol in enumerate(component_volumes_m3):
        print(f"  {component_names[i]}: {vol:.4f} m³")
      print(f"  Suma de volúmenes de componentes: {sum(component_volumes_m3):.4f} m³")
    except ValueError as e:
      print(f"Error al calcular volúmenes con volumen_absoluto: {e}")
  else:
    print("No hay suficientes datos para calcular volúmenes de componentes.")

  return final_masses, warning_message
```

### **Función `calcular_costo`**
Esta función calcula el costo total de todos los materiales necesarios para la dosificación del concreto, sumando el producto de la masa de cada componente por su precio unitario por kilogramo.

```python
def calcular_costo(quantities_kg: dict, unit_prices_per_kg: dict) -> float:
  """
  Calcula el costo total de una lista de materiales.

  Args:
    quantities_kg: Diccionario con las cantidades (masas en kg) de cada material.
                   Ejemplo: {'cement': 175, 'water': 87.5}
    unit_prices_per_kg: Diccionario con los precios unitarios por kilogramo (costo/kg)
                        de cada material. Ejemplo: {'cement': 0.10, 'water': 0.001}

  Returns:
    Costo total (float).
    Retorna 0.0 si no hay cantidades o precios válidos para calcular.
  """
  quantities_kg = _validate_and_convert_dict_values(quantities_kg, 'quantities_kg')
  unit_prices_per_kg = _validate_and_convert_dict_values(unit_prices_per_kg, 'unit_prices_per_kg')

  total_cost = 0.0

  for component, mass in quantities_kg.items():
    if component in unit_prices_per_kg:
      total_cost += mass * unit_prices_per_kg[component]
    else:
      print(f"Advertencia: No se encontró precio unitario para '{component}'. No se incluirá en el cálculo de costos.")

  return total_cost
```

### **Función `generar_reporte`**
Esta función consolida y presenta un informe detallado de la dosificación del concreto. Muestra las masas finales, los volúmenes individuales de cada componente, el costo total y el costo promedio por metro cúbico, además de cualquier advertencia generada.

```python
def generar_reporte(
    total_volume_requested_m3: float,
    final_masses: dict,
    warning_message: str | None,
    total_cost: float,
    specific_gravities: dict
) -> None:
  """
  Genera un reporte formateado con los resultados de la dosificación y costos.

  Args:
    total_volume_requested_m3: Volumen total de concreto pedido en metros cúbicos (m³).
    final_masses: Diccionario con las masas finales (kg) de cada componente.
    warning_message: Mensaje de advertencia (str) si la relación a/c excede 0.65, o None.
    total_cost: Costo total de los materiales.
    specific_gravities: Diccionario con las gravedades específicas de cada componente
                        para recalcular y mostrar los volúmenes individuales.
  """
  total_volume_requested_m3 = _validate_and_convert_numeric(total_volume_requested_m3, 'total_volume_requested_m3')
  final_masses = _validate_and_convert_dict_values(final_masses, 'final_masses')
  # warning_message doesn't need validation as it can be None or string
  total_cost = _validate_and_convert_numeric(total_cost, 'total_cost')
  specific_gravities = _validate_and_convert_dict_values(specific_gravities, 'specific_gravities')

  print("\n" + "="*40)
  print("           REPORTE DE DOSIFICACIÓN DE CONCRETO")
  print("="*40)

  print(f"\nVolumen total solicitado: {total_volume_requested_m3:.2f} m³")

  print("\n--- Masas Finales (kg) ---")
  for componente, masa in final_masses.items():
    print(f"  {componente.capitalize()}: {masa:.2f} kg")

  print("\n--- Volúmenes Individuales (m³) ---")
  masses_for_vol_calc = []
  sgs_for_vol_calc = []
  component_names = []

  ordered_components = ['cement', 'water', 'sand', 'gravel'] # Mantener el orden

  for component in ordered_components:
    if component in final_masses and component in specific_gravities:
      masses_for_vol_calc.append(final_masses[component])
      sgs_for_vol_calc.append(specific_gravities[component])
      component_names.append(component)

  if masses_for_vol_calc and sgs_for_vol_calc:
    try:
      component_volumes_m3 = volumen_absoluto(masses_for_vol_calc, sgs_for_vol_calc)
      for i, vol in enumerate(component_volumes_m3):
        print(f"  {component_names[i].capitalize()}: {vol:.4f} m³")
      print(f"  Volumen total aparente de componentes: {sum(component_volumes_m3):.4f} m³")
    except ValueError as e:
      print(f"Error al calcular volúmenes en el reporte: {e}")
  else:
    print("  No hay datos suficientes para calcular volúmenes de componentes.")

  # Calcular costo por m³
  cost_per_m3 = total_cost / total_volume_requested_m3 if total_volume_requested_m3 > 0 else 0.0

  print("\n--- Costos ---")
  print(f"  Costo Total de Materiales: ${total_cost:.2f}")
  print(f"  Costo Promedio por Metro Cúbico: ${cost_per_m3:.3f}/m³")

  if warning_message:
    print("\n--- Advertencia ---")
    print(f"  {warning_message}")

  print("="*40)
```

### **Programa Principal Interactivo de Dosificación de Concreto**
Esta celda ejecuta el programa principal, solicitando al usuario que ingrese interactivamente los parámetros de diseño del concreto, las masas unitarias, los precios y las gravedades específicas. Luego, utiliza las funciones definidas previamente para dosificar el concreto, calcular los costos y generar un reporte completo.

```python
print("\n" + "#"*50)
print("##### PROGRAMA PRINCIPAL DE DOSIFICACIÓN DE CONCRETO INTERACTIVO #####")
print("#"*50)

# --- 1. Definición de Parámetros de Diseño (Entrada del Usuario) ---

print("\n--- Ingrese los Parámetros del Concreto ---")
fc_deseado_main = float(input("Ingrese la resistencia a la compresión deseada (f'c en MPa, ej. 28): "))
volumen_total_pedido_main = float(input("Ingrese el volumen total de concreto a dosificar (en m³, ej. 1.5): "))
relacion_ac_diseno_main = float(input("Ingrese la relación Agua/Cemento de diseño (ej. 0.55): "))

print("\n--- Ingrese las Masas Unitarias por m³ (kg/m³) ---")
masas_unitarias_ejemplo_main = {}
components = ['cement', 'water', 'sand', 'gravel']
for comp in components:
    mass = float(input(f"Ingrese la masa unitaria de {comp} (kg/m³, ej. 400 para cemento): "))
    masas_unitarias_ejemplo_main[comp] = mass

print("\n--- Ingrese los Precios Unitarios de los Materiales (por kg) ---")
unit_prices_example_main = {}
for comp in components:
    price = float(input(f"Ingrese el precio unitario de {comp} (solo número, ej. 0.18 para cemento): "))
    unit_prices_example_main[comp] = price

print("\n--- Ingrese las Gravedades Específicas para el Reporte ---")
specific_gravities_main = {}
for comp in components:
    sg = float(input(f"Ingrese la gravedad específica de {comp} (ej. 3.15 para cemento): "))
    specific_gravities_main[comp] = sg

print("\n--- Parámetros de Entrada Confirmados ---")
print(f"Resistencia deseada (f'c): {fc_deseado_main} MPa")
print(f"Volumen total solicitado: {volumen_total_pedido_main} m³")
print(f"Relación Agua/Cemento de diseño: {relacion_ac_diseno_main}")
print(f"Masas Unitarias por m³: {masas_unitarias_ejemplo_main}")
print(f"Precios Unitarios por kg: {unit_prices_example_main}")
print(f"Gravedades Específicas: {specific_gravities_main}")

# --- 2. Dosificación del Concreto ---
print("\n--- Ejecutando Dosificación del Concreto ---")
try:
  masas_finales_main, advertencia_main = dosificar_concreto(
      fc_deseado_main,
      volumen_total_pedido_main,
      relacion_ac_diseno_main,
      masas_unitarias_ejemplo_main
  )
  print("Dosificación completada.")
except TypeError as e:
  print(f"Error en la dosificación debido a tipos de datos: {e}")
  masas_finales_main = {}
  advertencia_main = f"Error crítico: {e}"

# --- 3. Cálculo de Costos ---
print("\n--- Ejecutando Cálculo de Costos ---")
if masas_finales_main:
  try:
    total_cost_main = calcular_costo(masas_finales_main, unit_prices_example_main)
    print("Cálculo de costos completado.")
  except TypeError as e:
    print(f"Error en el cálculo de costos debido a tipos de datos: {e}")
    total_cost_main = 0.0
else:
  total_cost_main = 0.0
  print("No se pudo calcular el costo debido a errores previos en la dosificación.")

# --- 4. Generación del Reporte Final ---
print("\n--- Generando Reporte Final ---")
if masas_finales_main:
  try:
    generar_reporte(
        volumen_total_pedido_main,
        masas_finales_main,
        advertencia_main,
        total_cost_main,
        specific_gravities_main
    )
    print("Reporte generado exitosamente.")
  except TypeError as e:
    print(f"Error al generar el reporte debido a tipos de datos: {e}")
else:
  print("No se pudo generar el reporte debido a errores previos en la dosificación.")

print("\n" + "#"*50)
print("##### FIN DEL PROGRAMA PRINCIPAL #####")
print("#"*50)
```

1.  **¿Qué ventaja tiene descomponer el problema en funciones individuales versus resolver todo en un solo bloque de código?**
    Descomponer el problema en funciones individuales (modularización) ofrece varias ventajas:
    *   **Reusabilidad:** Las funciones pueden ser reutilizadas en diferentes partes del programa o en otros proyectos, evitando la duplicación de código.
    *   **Mantenibilidad:** Es más fácil depurar y modificar una pequeña función específica que buscar un error en un bloque de código grande. Los cambios en una función no afectan necesariamente a otras, siempre y cuando la interfaz (entradas/salidas) se mantenga consistente.
    *   **Legibilidad y Comprensión:** El código se vuelve más fácil de leer y entender, ya que cada función tiene una responsabilidad clara y específica. Esto mejora la colaboración en equipos.
    *   **Testing:** Permite probar cada componente (función) de forma aislada, lo que facilita la identificación y corrección de errores.
    *   **Abstracción:** Oculta los detalles de implementación, permitiendo a los usuarios de la función concentrarse en *qué* hace, no *cómo* lo hace.

2.  **¿Por qué es útil que una función retorne una tupla `(resultado, advertencia)` en lugar de solo el resultado?**
    Retornar una tupla `(resultado, advertencia)` es útil por las siguientes razones:
    *   **Comunicación de Estado:** Permite que la función no solo devuelva el resultado principal de su operación, sino también información adicional sobre el estado de la ejecución. Una advertencia indica que la operación se completó, pero bajo ciertas condiciones que podrían requerir atención o ser subóptimas, sin llegar a ser un error crítico que detenga el programa.
    *   **Flexibilidad en el Manejo:** El código que llama a la función puede decidir cómo manejar la advertencia (mostrarla, ignorarla, registrarla) sin que la función imponga una interrupción o un comportamiento fijo.
    *   **Separación de Preocupaciones:** Se distingue entre un resultado exitoso (incluso con peculiaridades) y una falla que impediría obtener un resultado válido (que se manejaría con excepciones).

3.  **Si el cliente (ingeniero civil) pide cambiar la relación a/c máxima de 0.65 a 0.60, ¿en qué parte del código haces el cambio y por qué no necesitas modificar todo el script?**
    El cambio se haría únicamente en la función `dosificar_concreto`. Específicamente, en la línea donde se compara `ac_ratio_calculated` con el valor límite (`0.65`). Simplemente se modificaría `0.65` a `0.60`.

    No es necesario modificar todo el script gracias a la modularización. La lógica para la validación de la relación agua/cemento está encapsulada dentro de `dosificar_concreto`. Las otras funciones (`volumen_absoluto`, `calcular_costo`, `generar_reporte`) y el programa principal llaman a `dosificar_concreto` sin necesidad de conocer los detalles internos de esa validación. Solo necesitarían saber que `dosificar_concreto` retorna una advertencia si la condición se cumple, pero no el valor específico de la condición.
