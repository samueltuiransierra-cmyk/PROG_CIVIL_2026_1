````python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# 1. Solicitar al usuario el número de plumas
plumas = float(input("Ingrese el número de plumas: "))

# 2. Definir constantes del problema
# Valor de una pluma en litros por segundo
CAUDAL_PLUMA = 0.025
# Segundos en un día
SEGUNDOS_DIA = 24 * 60 * 60 # 86400 segundos

# 3. Calcular el caudal total
caudal_total = plumas * CAUDAL_PLUMA

# 4. Calcular el consumo diario
consumo_diario = caudal_total * SEGUNDOS_DIA

# 5. Calcular el consumo mensual
# Se asume un mes de 30 días, como se indica en el problema
consumo_mensual = consumo_diario * 30

# 6. Clasificar el consumo
nivel_consumo = ""
if consumo_diario < 1000:
    nivel_consumo = "Bajo"
elif 1000 <= consumo_diario <= 3000:
    nivel_consumo = "Medio"
else:
    nivel_consumo = "Alto"

# 7. Mostrar resultados
print(f"\n--- Resultados del consumo ---")
print(f"Número de plumas: {plumas}")
print(f"Caudal total: {caudal_total:.4f} litros/segundo")
print(f"Consumo diario: {consumo_diario:.2f} litros/día")
print(f"Consumo mensual: {consumo_mensual:.2f} litros/mes (considerando 30 días)")
print(f"Clasificación del consumo: {nivel_consumo}")
````

## 8. Extensión opcional

Si desea profundizar, puede ampliar el programa para que también permita:

- calcular el consumo semanal;
- comparar el consumo de varios hogares;
- estimar el costo del agua si se conoce una tarifa por litro o por metro cúbico;
- convertir litros a metros cúbicos;
- presentar los resultados en forma tabular.

> Estas extensiones no reemplazan la solución base; son una ampliación.

````python
import pandas as pd

# ----------------------------------------------------------
# Extensión Opcional: Comparar Consumo de Varios Hogares
# ----------------------------------------------------------

print("--- Comparación de Consumo entre Hogares ---")

num_hogares = int(input("¿Cuántos hogares desea comparar? "))

# Definir constantes del problema (redefinidas para que la celda sea lo más independiente posible)
CAUDAL_PLUMA = 0.025
SEGUNDOS_DIA = 24 * 60 * 60 # 86400 segundos
LITROS_POR_METRO_CUBICO = 1000

resultados_hogares = []

for i in range(num_hogares):
    print(f"\n--- Datos para el Hogar {i+1} ---")
    plumas = float(input(f"Ingrese el número de plumas para el Hogar {i+1}: "))
    tarifa_m3 = float(input(f"Ingrese la tarifa por metro cúbico para el Hogar {i+1} (ej. 1.50): "))

    # 1. Calcular el caudal total
    caudal_total = plumas * CAUDAL_PLUMA

    # 2. Calcular el consumo diario
    consumo_diario = caudal_total * SEGUNDOS_DIA

    # 3. Calcular el consumo semanal y mensual
    consumo_semanal = consumo_diario * 7
    consumo_mensual = consumo_diario * 30

    # 4. Convertir a metros cúbicos
    consumo_diario_m3 = consumo_diario / LITROS_POR_METRO_CUBICO
    consumo_semanal_m3 = consumo_semanal / LITROS_POR_METRO_CUBICO
    consumo_mensual_m3 = consumo_mensual / LITROS_POR_METRO_CUBICO

    # 5. Calcular el costo del agua
    costo_diario = consumo_diario_m3 * tarifa_m3
    costo_semanal = consumo_semanal_m3 * tarifa_m3
    costo_mensual = consumo_mensual_m3 * tarifa_m3

    # 6. Clasificar el consumo
    nivel_consumo = ""
    if consumo_diario < 1000:
        nivel_consumo = "Bajo"
    elif 1000 <= consumo_diario <= 3000:
        nivel_consumo = "Medio"
    else:
        nivel_consumo = "Alto"

    resultados_hogares.append({
        'Hogar': i+1,
        'Plumas': plumas,
        'Tarifa m3': tarifa_m3,
        'Caudal Total (L/s)': round(caudal_total, 4),
        'Consumo Diario (L)': round(consumo_diario, 2),
        'Consumo Diario (m³)': round(consumo_diario_m3, 2),
        'Costo Diario': round(costo_diario, 2),
        'Consumo Semanal (L)': round(consumo_semanal, 2),
        'Consumo Semanal (m³)': round(consumo_semanal_m3, 2),
        'Costo Semanal': round(costo_semanal, 2),
        'Consumo Mensual (L)': round(consumo_mensual, 2),
        'Consumo Mensual (m³)': round(consumo_mensual_m3, 2),
        'Costo Mensual': round(costo_mensual, 2),
        'Nivel Consumo': nivel_consumo
    })

# Mostrar resultados en una tabla
df_comparacion = pd.DataFrame(resultados_hogares)
print("\n--- Tabla Comparativa de Consumo por Hogar ---")
display(df_comparacion)
````

```python
CAUDAL_PLUMA = 0.025  # Valor de una pluma de agua en litros por segundo.
SEGUNDOS_DIA = 24 * 60 * 60 # Número de segundos en un día (86400 segundos).
LITROS_POR_METRO_CUBICO = 1000 # Factor de conversión de litros a metros cúbicos.

# Lista para almacenar los resultados calculados para cada hogar.
# Cada elemento de esta lista será un diccionario con los datos de un hogar.
resultados_hogares = []

# Bucle principal para iterar sobre cada hogar a comparar.
for i in range(num_hogares):
    print(f"\n--- Datos para el Hogar {i+1} ---")
    
    # Solicitar el número de plumas para el hogar actual.
    # Se utiliza float() para permitir valores decimales.
    plumas = float(input(f"Ingrese el número de plumas para el Hogar {i+1}: "))
    
    # Solicitar la tarifa por metro cúbico para el hogar actual.
    tarifa_m3 = float(input(f"Ingrese la tarifa por metro cúbico para el Hogar {i+1} (ej. 1.50): "))

    # 1. Calcular el caudal total de agua para el hogar actual.
    # Caudal Total (L/s) = Número de plumas * Caudal por pluma.
    caudal_total = plumas * CAUDAL_PLUMA

    # 2. Calcular el consumo diario en litros.
    # Consumo Diario (L) = Caudal Total (L/s) * Segundos en un día.
    consumo_diario = caudal_total * SEGUNDOS_DIA

    # 3. Calcular el consumo semanal y mensual en litros.
    # Se asume una semana de 7 días y un mes de 30 días, según el planteamiento.
    consumo_semanal = consumo_diario * 7
    consumo_mensual = consumo_diario * 30

    # 4. Convertir los consumos de litros a metros cúbicos (m³).
    # 1 m³ = 1000 Litros.
    consumo_diario_m3 = consumo_diario / LITROS_POR_METRO_CUBICO
    consumo_semanal_m3 = consumo_semanal / LITROS_POR_METRO_CUBICO
    consumo_mensual_m3 = consumo_mensual / LITROS_POR_METRO_CUBICO

    # 5. Calcular el costo del agua para los consumos diario, semanal y mensual.
    # Costo = Consumo (m³) * Tarifa por m³.
    costo_diario = consumo_diario_m3 * tarifa_m3
    costo_semanal = consumo_semanal_m3 * tarifa_m3
    costo_mensual = consumo_mensual_m3 * tarifa_m3

    # 6. Clasificar el nivel de consumo (Bajo, Medio, Alto) basado en el consumo diario en litros.
    nivel_consumo = "" # Inicializar la variable de nivel de consumo.
    if consumo_diario < 1000:
        nivel_consumo = "Bajo" # Si el consumo diario es menor a 1000 L.
    elif 1000 <= consumo_diario <= 3000:
        nivel_consumo = "Medio" # Si el consumo diario está entre 1000 L y 3000 L (inclusive).
    else:
        nivel_consumo = "Alto" # Si el consumo diario es mayor a 3000 L.

    # Almacenar todos los resultados calculados para el hogar actual en un diccionario.
    # Se usa round() para formatear los números a un número específico de decimales para la presentación.
    resultados_hogares.append({
        'Hogar': i+1,
        'Plumas': plumas,
        'Tarifa m3': tarifa_m3,
        'Caudal Total (L/s)': round(caudal_total, 4),
        'Consumo Diario (L)': round(consumo_diario, 2),
        'Consumo Diario (m³)': round(consumo_diario_m3, 2),
        'Costo Diario': round(costo_diario, 2),
        'Consumo Semanal (L)': round(consumo_semanal, 2),
        'Consumo Semanal (m³)': round(consumo_semanal_m3, 2),
        'Costo Semanal': round(costo_semanal, 2),
        'Consumo Mensual (L)': round(consumo_mensual, 2),
        'Consumo Mensual (m³)': round(consumo_mensual_m3, 2),
        'Costo Mensual': round(costo_mensual, 2),
        'Nivel Consumo': nivel_consumo
    })

# Después de procesar todos los hogares, convertir la lista de diccionarios en un DataFrame de pandas.
# Esto proporciona una estructura tabular para una fácil visualización y comparación.
df_comparacion = pd.DataFrame(resultados_hogares)

# Mostrar la tabla comparativa final.
print("\n--- Tabla Comparativa de Consumo por Hogar ---")
display(df_comparacion)
```
